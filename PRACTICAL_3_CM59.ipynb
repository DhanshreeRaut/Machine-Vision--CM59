```python
import os
import glob
import cv2
import matplotlib.pyplot as plt

# Path to training images
image_folder = "car/train/images"

# Get all jpg and png images
image_list = glob.glob(os.path.join(image_folder, "*.jpg"))
image_list += glob.glob(os.path.join(image_folder, "*.png"))

print("Total Images:", len(image_list))

# Select first image automatically
reference_image = image_list[0]

print("Selected Image:", reference_image)
```

```python
img = cv2.imread(reference_image)

if img is None:
    print("Image not found!")
else:
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(6,6))
    plt.imshow(img)
    plt.axis("off")
    plt.title("Reference Image")
    plt.show()
```
